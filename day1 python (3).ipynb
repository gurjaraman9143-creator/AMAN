```python

import pandas as pd
import numpy as np

# DataFrame from Dictionary of Columns

Student = pd.DataFrame({
    "name": ["Arnav", "diya", "kabir", "Meera", "Rohan", "Sara"],
    "Branch": ["CSE", "CSE", "ECE", "ECE", "CSE", "ME"],
    "marks": [88, 80, 70, np.nan, 60, 80],
    "hours": [4, 8, 5, 6, 4, 6]
})

print(Student)
```

```python
print("Shape :", Student.shape)
print("Columns:", Student.columns)
```

```python
print ("info:", Student.info) 
print("head:", Student.head(2))
```

```python
print(Student["marks"])
print(Student[["name","marks"]])
```

```python
print(Student.loc[0:2,"name"])
print(Student.iloc[0:2,0])
print(Student.iloc[0:2,0:3])
Student[Student["marks"]>80]
Student[Student["hours"]>1]
```

```python
print(Student[(Student["Branch"] == "CSE") | (Student["marks"] > 70)])
```

```python
print(Student[(Student["Branch"] == "CSE") & (Student["marks"] > 70)])
```

```python
df = Student.copy()
df["marks_pct"] = df["marks"] / 60
df
```

```python
df["passed"] = df["marks"] >=70
df
```

```python
df["grade"] = pd.cut(df["marks"],
                     bins=[0,40,50,60,70],
                     labels=["A", "B", "C", "D"])
df
```
