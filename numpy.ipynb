```python
# Indexing, slicing and boolean 

import numpy as np
a = np.array([12, 45, 7, 23, 56, 31, 8, 64, 19, 2])
print(a)
```

```python
last_three = a[-3:]
print(last_three)
```

```python
big = a[a > 20]
print(big)
```

```python
mid = a[(a >= 10) & (a <= 50)]
print(mid)
```

```python
every_second = a[::2]
print(every_second)
```

```python
n_big = np.sum(a > 20)
print(n_big)
```

```python
#Aggregations and the axis argument
import numpy as np
```

```python
scores = np.random.default_rng(3).integers(40, 101, size=(5, 4)).astype(float)
per_subject_mean = scores.mean(axis=0)
print(per_subject_mean)
```

```python
per_student_mean = scores.mean(axis=1)
print(per_student_mean)
```

```python
best_in_subject = scores.argmax(axis=0)
print(best_in_subject)
```

```python
top_student = per_student_mean.argmax()
print(top_student)
```

```python
overall_max = scores.max()
print(overall_max)
```

```python
 #Missing values: np.nan

v = np.array([4.0, 8.0, np.nan, 6.0, np.nan, 10.0])
```

```python
plain_mean = np.mean(v)
print(plain_mean)
```

```python
safe_mean = np.nanmean(v)
print(safe_mean)
```

```python
n_missing = np.isnan(v).sum()
print(n_missing)
```

```python
median = np.nanmedian(v)
v_filled = np.where(np.isnan(v), median, v)
print(v_filled)
```

```python
nan_equals_nan = np.nan == np.nan
print(nan_equals_nan)
```
